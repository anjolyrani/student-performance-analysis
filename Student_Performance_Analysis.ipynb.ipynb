```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from google.colab import files

uploaded = files.upload()
```

```python

file_name = next(iter(uploaded))

df = pd.read_csv(file_name)

print("Dataset loaded successfully!")
display(df.head(10))
```

```python
print("Number of rows and columns:", df.shape)

print("\nColumn names:")
print(df.columns.tolist())

print("\nData types and non-null values:")
df.info()

print("\nMissing values:")
print(df.isnull().sum())

print("\nNumerical columns:")
print(df.select_dtypes(include="number").columns.tolist())

print("\nCategorical columns:")
print(df.select_dtypes(include=["object", "category", "bool"]).columns.tolist())
```

```python
numeric_cols = df.select_dtypes(include="number").columns.tolist()

if numeric_cols:
    num_col = (
        "midterm_score"
        if "midterm_score" in numeric_cols
        else numeric_cols[0]
    )

    sns.histplot(data=df, x=num_col, bins=15, kde=True)
    plt.title(f"Distribution of {num_col}")
    plt.xlabel(num_col)
    plt.ylabel("Number of observations")
    plt.tight_layout()
    plt.show()
else:
    print("No numerical columns found.")
```

```python

cat_cols = df.select_dtypes(
    include=["object", "category", "bool"]
).columns.tolist()

if cat_cols and numeric_cols:
    cat_col = cat_cols[0]
    num_col = (
        "total_score"
        if "total_score" in numeric_cols
        else numeric_cols[0]
    )

    group_means = df.groupby(
        cat_col, observed=True
    )[num_col].mean().sort_values(ascending=False)

    group_means.plot(kind="bar")
    plt.title(f"Average {num_col} by {cat_col}")
    plt.xlabel(cat_col)
    plt.ylabel(f"Average {num_col}")
    plt.xticks(rotation=45, ha="right")
    plt.tight_layout()
    plt.show()

    print("Highest group:", group_means.idxmax())
    print("Lowest group:", group_means.idxmin())
else:
    print("Suitable categorical or numerical columns not found.")
```

```python

if len(numeric_cols) >= 2:
    x_col = (
        "study_hours_per_week"
        if "study_hours_per_week" in numeric_cols
        else numeric_cols[0]
    )

    y_col = (
        "total_score"
        if "total_score" in numeric_cols and "total_score" != x_col
        else next(c for c in numeric_cols if c != x_col)
    )

    sns.scatterplot(data=df, x=x_col, y=y_col)
    plt.title(f"{x_col} vs {y_col}")
    plt.xlabel(x_col)
    plt.ylabel(y_col)
    plt.tight_layout()
    plt.show()

    print("Correlation:", df[x_col].corr(df[y_col]))
else:
    print("At least two numerical columns are needed.")
```

```python

if cat_cols and numeric_cols:
    cat_col = cat_cols[0]
    num_col = (
        "total_score"
        if "total_score" in numeric_cols
        else numeric_cols[0]
    )

    sns.boxplot(data=df, x=cat_col, y=num_col)
    plt.title(f"Distribution of {num_col} by {cat_col}")
    plt.xlabel(cat_col)
    plt.ylabel(num_col)
    plt.xticks(rotation=45, ha="right")
    plt.tight_layout()
    plt.show()
else:
    print("Suitable columns not found.")
```

```python

if len(numeric_cols) >= 2:
    correlation = df[numeric_cols].corr()

    plt.figure(figsize=(10, 7))
    sns.heatmap(
        correlation,
        annot=True,
        cmap="coolwarm",
        fmt=".2f",
        vmin=-1,
        vmax=1
    )
    plt.title("Correlation Between Numerical Variables")
    plt.tight_layout()
    plt.show()
else:
    print("At least two numerical columns are needed.")
```

```python

import pandas as pd
import os

files = [f for f in os.listdir('/content') if f.endswith('.csv')]
print(files)
```

```python

df = pd.read_csv('/content/student_performance_exploration_v2 - student_performance_exploration_v2.csv')

df.head()
```

```python

print("Dataset Shape:", df.shape)
print("\nColumn Names:")
print(df.columns.tolist())

print("\nDataset Information:")
df.info()
```

```python
df.isnull().sum()
```

```python
df_clean = df.copy()

# Fill missing numeric values with their median
numeric_cols = [
    "study_hours_per_week",
    "attendance_percentage",
    "sleep_hours_per_night",
    "assignment_score"
]

for col in numeric_cols:
    df_clean[col] = df_clean[col].fillna(
        df_clean[col].median()
    )

print("Missing values after cleaning:")
print(df_clean.isnull().sum())
```

```python
print("Duplicate rows:", df_clean.duplicated().sum())
print("Duplicate student IDs:", df_clean["student_id"].duplicated().sum())
```

```python
print("Average study hours:",
      df_clean["study_hours_per_week"].mean())

# Average attendance
print("Average attendance:",
      df_clean["attendance_percentage"].mean())

# Average total score
print("Average total score:",
      df_clean["total_score"].mean())

# Highest total score
print("Highest total score:",
      df_clean["total_score"].max())

# Lowest total score
print("Lowest total score:",
      df_clean["total_score"].min())
```

```python

print(df_clean["passed_course"].value_counts())

print("\nPass percentage:")
print(
    df_clean["passed_course"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)
```

```python

department_analysis = df_clean.groupby("department").agg(
    student_count=("student_id", "count"),
    average_score=("total_score", "mean"),
    average_attendance=("attendance_percentage", "mean"),
    average_study_hours=("study_hours_per_week", "mean")
).round(2)

department_analysis
```

```python

import matplotlib.pyplot as plt

df_clean.groupby("department")["total_score"].mean().plot(
    kind="bar",
    figsize=(8, 5)
)

plt.title("Average Total Score by Department")
plt.xlabel("Department")
plt.ylabel("Average Total Score")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
df_clean.to_csv(
    "/content/cleaned_student_performance.csv",
    index=False
)

print("Cleaned dataset saved successfully!")
```
